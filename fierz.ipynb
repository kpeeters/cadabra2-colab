# Fierz transformations

This notebook verifies a Fierz identity for Majorana spinors in eleven dimensions, used in B. de Wit, K. Peeters and J. Plefka, "Superspace geometry for supermembrane backgrounds", Nucl. Phys. B532 (1998) 99, arXiv:hep-th/9803209. It is also discussed in the original Cadabra paper, arXiv:hep-th/0701238.

We first declare the properties of all symbols. 

```python
{\mu,\nu,\rho}::Indices(curved, position=fixed).
{m,n,p,q,r,s,t,u,v}::Indices(flat, position=independent).
{m,n,p,q,r,s,t,u,v}::Integer(0..10).
T^{#{\mu}}::AntiSymmetric.
\psi_{\mu}::SelfAntiCommuting.
\psi_{\mu}::Spinor(dimension=11, type=Majorana).
\theta::Spinor(dimension=11, type=Majorana).
\epsilon::Spinor(dimension=11, type=Majorana).
{\theta,\epsilon,\psi_{\mu}}::AntiCommuting.
\bar{#}::DiracBar.
\delta^{m n}::KroneckerDelta.
\Gamma^{#{m}}::GammaMatrix(metric=\delta).
```

The expression we want to transform is the following (the anti-symmetric tensor $T^{\mu\nu\rho}$ is used to implement anti-symmetrisation in the curved indices). We want to use a Fierz transformation so that the spinors are paired as $(\bar{\theta}, \epsilon)$ and $(\bar{\psi_{\mu}}, \psi_{\rho})$. 

```python
obj:= T^{\mu\nu\rho} e_{\nu}^{s}
     \bar{\theta} \Gamma^{r s} \psi_{\rho}
     \bar{\psi_{\mu}} \Gamma^{r} \epsilon;
```

The reorganisation of the spinors is done with a single command, which states the spinors in the order in which we want to have them.

```python
fierz(_, $\theta, \epsilon, \psi_{\mu}, \psi_{\rho}$ );
```

Now we can still rewrite this in terms of fully anti-symmetrised gamma matrix products. The only thing we need to do is to repeatedly write gamma products in terms of fully-antisymmetrised products using `join_gamma`, and then distribute the product and eliminate contractions.

```python
converge(obj):
   join_gamma(_)
   distribute(_)
   eliminate_kronecker(_)
canonicalise(_)
rename_dummies(_);
```

This agrees with (5.86) in K. Peeters, "Geometrical aspects of supersymmetry" (1998) University of Utrecht.
