# Gamma matrix algebra (Clifford algebras)

Cadabra can do gamma matrix algebra in an arbitrary number of dimensions. The example below shows how to work out 
the particular gamma matrix contraction

$$
\Gamma_{s r} \Gamma_{r l} \Gamma_{k m} \Gamma_{m s}\,,
$$

which is not impossible to do by hand, but certainly much nicer with Cadabra. As always, we first need to declare
the symbols we are going to use. 

```python
{s,r,l,k,m,n}::Indices(vector);
{s,r,l,k,m,n}::Integer(0..d-1);
\Gamma_{#}::GammaMatrix(metric=\delta);
\delta_{m n}::KroneckerDelta;
```

After every step in which we work out the product of two generalised gamma matrices, we need a number of algorithms
to simplify the result. We will use the mechanism of defining a `post_process` function, which gets called after
every Cadabra function call. Note how every algorithm that is called in here acts on the same expression; you do not have to 
do things like `ex = sort_product(ex)`.

```python
def post_process(ex):
	sort_product(ex)
	eliminate_kronecker(ex)
	canonicalise(ex)
	collect_terms(ex)
```
