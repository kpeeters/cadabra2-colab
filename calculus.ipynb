# Basic calculus

Because Cadabra started as a tool for tensor manipulations, it has its own logic to manipulate mathematical expressions that contain tensors. However, it can use SymPy algorithms to manipulate the scalar parts of these expressions, and you can therefore combine the power of both in the same unified LaTeX notation.

Let's see this at the level of some simple computations. First, a Taylor series expansion done using SymPy, fed into Cadabra plotting functionality. We start from $\sin(x)$,

```python
ex:=\sin(x);
```

SymPy's calculus functionality is imported using a Cadabra package,

```python
from cdb.sympy.calculus import *
```

You can now call SymPy functions on Cadabra `ex` expressions; here is a series expansion to 10th and 20th order:

```python
s10 = series(ex, $x$, 0, 10);
```

```python
s20=series(ex, $x$, 0, 20);
```

Cadabra's plotting functionality resides in another package, from which we import `plot`, which we then use to plot the function and its two approximations,

```python
from cdb.graphics.plot import *
```

```python
plot([ex, s10, s20], ($x$,0,10), yrange=(-4,4), legend=["sin(x)", "O(10)", "O(20)"]);
```

You can differentiate just like in SymPy:

```python
diff($sin(x)$, $y$);
```

```python
diff($10 + 3*x + 4*y + 10*x**2 + x**9$, $x$);
```

```python
diff($exp(-1/2  x**2 )$, $x$);
```

or integrate

```python
ex:=\int{\log{x}**2}{x};
```

```python
integrate($\log{x}**2$, $x$);
```
