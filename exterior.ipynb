# Differential forms

Cadabra can handle the calculus of differential forms. Differential forms are declared by attaching
the property [`DifferentialForm`](https://cadabra.science/manual/DifferentialForm.html) to an object, as in the example below, where we declare
four forms, with degree 0 to 3, and one form with symbolic degree $p$.

```python
A0::LaTeXForm("A^{(0)}").
A1::LaTeXForm("A^{(1)}").
A2::LaTeXForm("A^{(2)}").
A3::LaTeXForm("A^{(3)}").
Bp::LaTeXForm("B^{(p)}").
A0::DifferentialForm(degree=0);
A1::DifferentialForm(degree=1);
A2::DifferentialForm(degree=2);
A3::DifferentialForm(degree=3);
Bp::DifferentialForm(degree=p);
```

The exterior product (or wedge product) of forms is denoted with a `^` symbol (make sure to put a space
before and after this symbol to avoid confusion with a superscript).

```python
ex:=A1 ^ A2 + A2 ^ A1;
```

```python
sort_product(_);
```

```python
ex:=A1 ^ A2 ^ A1;
sort_product(_);
```

```python
ex:= A3 ^ A2 ^ A1 ^ A2;
sort_product(_);
```

An exterior derivative can be declared by using the [`ExteriorDerivative`](https://cadabra.science/manual/ExteriorDerivative.html) property. Applying an exterior
derivative twice produces zero, and it obeys the product rule taking into account the degree of differential forms.

```python
d{#}::ExteriorDerivative;
d{#}::LaTeXForm("{\rm d}").
```

```python
ex2:= d{ @(ex) };
```

```python
product_rule(_);
```

```python
ex:= d{ A1 ^ A2 ^ d{ A3 } };
```

```python
product_rule(_);
```

```python
sort_product(_);
```

Two forms in a product can only be swapped around if one of their degrees is zero:

```python
ex4:= A2 A0 + A0 A2;
sort_product(_);
```

In a wedge product, forms can be swapped around taking into account their degree:

```python
ex5:= A3 ^ A1 - A1 ^ A3 + A2 ^ A2 ^ A0;
sort_product(_);
```

Differential forms can have tensor indices, e.g. for vielbeine or spin connections. In the example below the spin connection is a one-form:

```python
{a,b,c}::Indices;
{e^{a}, \omega^{a}_{b}}::DifferentialForm(degree=1);
```

```python
ex:=d{e^{a}} = - \omega^{a}_{b} ^ e^{b};
```

```python
cv:= d{ @(ex) };
```

```python
distribute(cv);
product_rule(cv);
```

```python
substitute(cv, ex);
```

```python
rename_dummies(cv);
```
