# Automatic multi-term simplification

This notebook illustrates how the automatic multi-term simplification algorithm [`meld`](https://cadabra.science/manual/meld.html) can
be used in a `post_process` function to simplify expressions immediately. For simpler computations,
or computations that are already working and for which you do not want to show these simplifications
at every step, this can be useful.

```python
def post_process(ex):
    meld(ex)
    collect_terms(ex)
```

```python
R_{a b c d}::RiemannTensor;
```

By the first Bianchi identity, the sum of cyclic permutations of three indices is zero:

```python
ex:=R_{a b c d} + R_{b c a d} + R_{c a b d};
```

Let's define a fully anti-symmetric tensor $A^{a b c}$ and a fully symmetric tensor $S^{a b c}$:

```python
A^{a b c}::AntiSymmetric;
```

```python
S_{a b c}::Symmetric;
```

Contracting a fully anti-symmetric tensor with the Riemann tensor leads to the first Bianchi identity again, so should vanish:

```python
ex:=R_{a b c d} A^{a b c};
```

```python
ex:= (A^{a b c} + A^{c a b}) S_{a b c};
```

Note that the post-processing function only applies the meld algorithm once, so the best it could do was to combine
the two anti-symmetric tensors. If you apply meld again, it sees the zero:

```python
meld(ex, repeat=True);
```

You could have achieved the above by instead using a `converge` construction inside the post-processing function.

Some other assorted examples now. Meld will automatically factor out scalar expressions:

```python
ex:= a A^{a b c} + b A^{b a c};
```

Not all expressions quadratic in the Riemann tensor are independent:

```python
ex:=R_{a b c d} R_{a b c d} + R_{a b c d} R_{a c b d};
```
